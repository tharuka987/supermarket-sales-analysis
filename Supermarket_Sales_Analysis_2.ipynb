```python
import pandas as pd

df = pd.read_csv('SuperMarket Analysis.csv')
print(df.shape)
print(df.head(3))
print(df.columns.tolist())
```

```python
# 1. total Sales by City
print("--- Sales by City ---")
print(df.groupby('City')['Sales'].sum().sort_values(ascending=False))

# 2. total Sales by Product line
print("\n--- Sales by Product line ---")
print(df.groupby('Product line')['Sales'].sum().sort_values(ascending=False))

# 3. average bill by Customer type
print("\n--- average bill by Customer type ---")
print(df.groupby('Customer type')['Sales'].mean())

# 4. most used Payment method
print("\n--- Payment method count ---")
print(df['Payment'].value_counts())
```

```python
import matplotlib.pyplot as plt

# graph 1 - Sales by city
plt.figure(figsize=(8,4))
df.groupby('City')['Sales'].sum().plot(kind='bar',color=['skyblue','lightgreen','salmon'])
plt.title('total sales by City')
plt.ylabel('sales')
plt.show()

# graph 2 - Sales by Product line
plt.figure(figsize=(10,4))
df.groupby('Product line')['Sales'].sum().sort_values().plot(kind='barh',color='orange')
plt.title('total Sales by Product line')
plt.xlabel('Sales')
plt.show()
```
