```python
import pandas as pd
print("project started!")
```

```python
data = {
    'product' : ['milk', 'bread', 'rice', 'eggs', 'soap', 'milk', 'bread', 'rice', 'eggs', 'soap'],
    'sales' : [120, 80, 250, 60, 90, 150, 95, 280, 70, 110],
    'city' : ['colombo', 'kandy', 'colombo', 'galle', 'kandy', 'colombo', 'galle', 'colombo', 'kandy', 'galle']
}

df = pd.DataFrame(data)
print(df)

```

```python
total = df['sales'].sum()
print("total sales:",total)

print(df.groupby('product')['sales'].sum())
```

```python
df.groupby('product')['sales'].sum().plot(kind = 'bar', color = 'orange')
```
